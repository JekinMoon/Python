### 문제 7
- 텐서 [[1, 2, 3], [4, 5, 6]]를 2개의 조각으로 나눠서 출력해보세요. (행 기준).

### 문제9
- 텐서 [[1, 2], [3, 4], [5, 6]]에서:
- dim=1 기준으로 크기 1씩 나누세요.
- 각각의 조각을 새로운 축에 쌓은 후 출력해보세요.

````python
t = torch.tensor([[1, 2], 
                  [3, 4], 
                  [5, 6]])

# 1. 열 기준으로 크기 1씩 나누기
t1= torch.split(t, 1, dim=1) 
print(t1)

# 2. 새로운 축에 쌓기
result = torch.stack(t1)
print(result)
````

### 문제10
- 다음 텐서를 활용해서 아래 문제를 해결해보세요
  ```
  a = torch.tensor([[1, 2], [3, 4]])
  b = torch.tensor([[5, 6], [7, 8]])
  ```

1. 두 텐서를 열 기준으로 연결하고 결과를 출력하세요.
2. 연결된 텐서를 크기 `(4, 2)`로 변환하세요.
3. 변환된 텐서를 2개의 조각으로 나누고, 각 조각을 0번 축으로 스택하여 출력하세요.


````python
a = torch.tensor([[1, 2], 
                  [3, 4]])
b = torch.tensor([[5, 6], 
                  [7, 8]])
# 1. 열 기준으로 병합
c = torch.cat([a,b], dim=1)
print(c)

# 2. 크기 (4, 2)로 변환
d = c.view(4,2)
print(d)

# 3. 2개의 조각으로 나누고
e1, e2 = torch.chunk(d,2,dim=0)
print(e1)
print(e2)

# 4. 0번 축으로 스택
f = torch.stack((e1,e2), dim=0)
print(f)
````
